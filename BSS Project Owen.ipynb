```python
# Récupération des données

# Les données sont contenues dans le dossier csv 
# r0X = 4 enregistrements abdominaux en fonction du temps (1 par colonne) avec X la patiente prise en charge + 1 enregistrement foetal direct
# qrs = 1 jeu d'annotations en fonction du temps, par X, tous dans la même colonne
# channels = métadonnées des enregistrements
fichier = Path("csv")

# Par exemple, 'r01' a 5 colonnes : t_s ; Direct_1, Abdomen_1, Abdomen_2, Abdomen_3, Abdomen_4
signaux = {f.stem: pd.read_csv(f) for f in sorted(fichier.glob("r*.csv"))}

# qrs : enregistrement ; sample ; ts
qrs = pd.read_csv(fichier / "qrs.csv")

# channels : enregistrement ; channel ; unit ; fs ; sensor_type ; prefiltering
channels = pd.read_csv(fichier / "channels.csv")

# La fréquence d'échantillonnage est la même pour tous les signaux, qu'ils soient Directs ou abdominaux
# Seules la fréquence d'échantillonnage des annotations varie, on règle ça plus loin
fs = channels["fs"].iloc[0]
```

```python
# On Découpe X01 en 5 fenêtres de 15 s, choisies dans les zones calmes de r01

duree = 15                          
debuts = [10, 30, 135, 255, 280]    
n = int(duree * fs)                 # nombre d'échantillons par fenêtre (fs est un flottant)

# Dictionnaire {t0: matrice 4 x n} 
fenetres = {}
for t0 in debuts:
    k0 = int(t0 * fs)               # colonne de X01 correspondant à l'instant t0
    fenetres[t0] = X01[:, k0 : k0 + n]

# QRS annotés de chaque fenêtre
qrs_r01 = qrs[qrs["record"] == "r01"]["t_s"].to_numpy()
qrs_fenetres = {t0: qrs_r01[(qrs_r01 >= t0) & (qrs_r01 < t0 + duree)] - t0 for t0 in debuts}

for t0 in debuts:
    print(f"{t0:3d}-{t0 + duree:3d} s : forme {fenetres[t0].shape}, {len(qrs_fenetres[t0])} QRS annotés")
```

```python
from scipy.signal import find_peaks
import random

enregistrement, t_debut, t_final = "r01", 0, 15 # Jouer ici sur les dates pour visualiser des fenêtes plus courtes
df = signaux[enregistrement]
window = df[(df["t_s"] >= t_debut) & (df["t_s"] < t_debut + t_final)]
qrs_window = qrs[(qrs["record"] == enregistrement) & (qrs["t_s"] >= t_debut) & (qrs["t_s"] < t_debut + t_final)]


signal_direct = window["Direct_1"]

list_peaks = find_peaks(signal_direct, height=None, threshold=None, distance =308, prominence=30, width=None, wlen=None, rel_height=0.5, plateau_size=None)[0]
# Constraint : maximum heartrate of 195

print("Number of detected peaks: " + str(len(list_peaks)))
print("Real number of heartbeats: " + str(len(qrs_window["t_s"])))
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 9), sharex=True)
axes[0].plot(window["t_s"], signal_direct, linewidth=0.5)
axes[0].vlines(qrs_window["t_s"], 0, 1, transform=axes[0].get_xaxis_transform(), color="red", linewidth=0.8, alpha=0.5)
axes[0].set_ylabel("Signal (µV)")
axes[0].set_title("Annotations")

axes[1].plot(window["t_s"], signal_direct, linewidth=0.5)
axes[1].vlines(window["t_s"][list_peaks+t_debut*fs], 0, 1, transform=axes[1].get_xaxis_transform(), color="green", linewidth=0.8, alpha=0.5)
axes[1].set_ylabel("Signal (µV)")
axes[1].set_title("Detected peaks")

axes[-1].set_xlabel("Time (s)")
plt.show()
```

```python
def error(annotations, detected_peaks, coeff = 1):
    annotations = np.array(annotations)
    detected_peaks = np.array(detected_peaks)
    nb_peaks = len(annotations)
    nb_detected_peaks = len(detected_peaks)
    if nb_detected_peaks==nb_peaks:
        average_error = sum(abs(annotations-detected_peaks))/nb_peaks
        return average_error
    else:
        total_error = 0
        for i in range(nb_detected_peaks*(nb_detected_peaks<=nb_peaks)+nb_peaks*(nb_detected_peaks>nb_peaks)):
            total_error += abs(detected_peaks[i]-annotations[i])
            i=i+1
        average_error = total_error/nb_peaks
        average_error += coeff*abs(nb_detected_peaks-nb_peaks)*annotations[nb_peaks-1]/nb_peaks #Add to the average error the number of peaks missed or created, weighted by the average time between two heartbeats, times a given penalty coefficient
        return average_error 
```

```python
# The coefficient is a coefficient for the penalty for an incorrect number of detected peaks. 
# A high value of the coefficient reduces the relative importance of the position of missed/incorrect peaks in the signal 
# Without penalty or with a low penalty, the error is significantly higher if a missed/incorrect peak is at the beginning of the signal
# Without a very high penalty, the position of peaks become almost irrelevant compared to their number 

for coeff in [0.5,1,2,4,8]:
    errors = np.zeros((100,10))
    for j in range(10):
        for i in range(100):
            detected_peaks = np.array(window["t_s"][list_peaks+t_debut*fs])
            annotations = qrs_window["t_s"]
            detected_peaks = np.delete(detected_peaks,random.sample([l for l in range(len(annotations))], k=j))
            errors[i,j] = error(annotations, detected_peaks,coeff)
            i=i+1
        j=j+1

    plt.figure()
    plt.plot(range(0,10), [np.mean(errors[:,j]) for j in range(0,10)],ls='-', c="blue")
    for j in range(10):
        plt.boxplot(errors[:,j], positions = [j])
    plt.title("Evolution of the error function with the number of missed peaks"+"\n Value of the penalty coefficient:"+str(coeff))
    plt.xlabel("Number of missed/incorrect peaks")
    plt.ylabel("Error")
    plt.show()
```
